```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_regression
import sklearn
import pandas as pd

sns.set(font_scale = 1.3)
np.random.seed(42)
np.set_printoptions(precision = 6, suppress = True)
```

```python
data, target = make_regression(n_samples=200,n_features=1,noise=20)
target = target ** 2

data[:5]
```

```python
target[:5]
```

```python
plt.figure(figsize=(8,6))
plt.title('Regresja liniowa')
plt.scatter(data, target, color = '#2100fa')
plt.legend()
plt.xlabel('cecha x')
plt.ylabel('target')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(data, target)


plot_data = np.arange(-3, 3, 0.01).reshape(-1, 1)

plt.figure(figsize=(8, 6))
plt.title('Regresja liniowa')
plt.plot(plot_data, regressor.predict(plot_data), c='red', label='regresja liniowa')
plt.scatter(data, target, label='dane')
plt.legend()
plt.xlabel('cecha x')
plt.ylabel('target')
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor

regressor = DecisionTreeRegressor()
regressor.fit(data, target)

plt.figure(figsize=(8, 6))
plt.title('Regresja drzew decyzyjnych')
plt.plot(plot_data, regressor.predict(plot_data), c='green', label=f'regresja drzew decyzyjnych')
plt.scatter(data, target, label='dane')
plt.legend()
plt.xlabel('cecha x')
plt.ylabel('target')
plt.show()
```

```python
y_true = 100 + 20 * np.random.randn(50)
y_true
```

```python
y_pred = y_true + 10 * np.random.randn(50)
y_pred
```

```python
results = pd.DataFrame({'y_true': y_true, 'y_pred': y_pred})
results.head()
```

```python
results['error'] = results['y_true'] - results['y_pred']
results['error_squared'] = results['error'] ** 2
results.head()
```

```python
print(f"MAE:{results['error'].abs().sum() / len(results):.4f}")
```

```python
print(f"MSE:{results['error_squared'].sum()/len(results):.4f}")
```

```python
print(f"RMSE:{np.sqrt(results['error_squared'].sum()/len(results)):.4f}")
```

```python
import plotly.graph_objects as go

def plot_regression_results(y_true, y_pred):

    results = pd.DataFrame({'y_true': y_true, 'y_pred': y_pred})
    min = results[['y_true', 'y_pred']].min().min()
    max = results[['y_true', 'y_pred']].max().max()

    fig = go.Figure(data=[go.Scatter(x=results['y_true'], y=results['y_pred'], mode='markers'),
                    go.Scatter(x=[min, max], y=[min, max])],
                    layout=go.Layout(showlegend=False, width=800,
                                     xaxis_title='y_true',
                                     yaxis_title='y_pred',
                                     title='Regresja: y_true vs. y_pred'))
    fig.show()

plot_regression_results(y_true, y_pred)
```

```python
from sklearn.metrics import mean_absolute_error

mean_absolute_error(y_true, y_pred)
```

```python
from sklearn.metrics import mean_squared_error

mean_squared_error(y_true, y_pred)
```

```python
np.sqrt(mean_squared_error(y_true, y_pred))
```

```python
from sklearn.metrics import max_error

max_error(y_true, y_pred)
```
